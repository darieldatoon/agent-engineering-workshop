# Part 1.1 The Agent Development Lifecycle

After this notebook you can name the four stages of the lifecycle, the LangSmith tools
for each, and where today's three parts fit.

## See it

<img src="../../assets/slides/adlc-5-on-langsmith.jpg" width="720">

One loop, four stages. **Build** with `deepagents`, `langchain`, `langgraph`, or
LangSmith Fleet. **Test** with datasets, evals, and experiments. **Deploy** on a runtime
with the Agent Server, sandboxes, and Context Hub. **Monitor** with tracing,
dashboards, online evals, and user feedback. **Govern** wraps the whole loop, and the
LLM Gateway is how every model call in this workshop is routed and controlled.

## Build it

Where each part of today fits:

| Part | Stages | What you do | LangSmith pieces |
| --- | --- | --- | --- |
| 1. LangGraph | Build + Test | Run and rebuild a baggage-claims **workflow**, then score it | Studio, tracing, datasets, evaluators, experiments |
| 2. Deep Agents | Build + Test | Build a rebooking **agent** and score its tool use | Studio, tracing, datasets, evaluators, experiments |
| 3. Deploy and monitor | Deploy + Monitor | Ship the agent and watch live traffic | Agent Server, tracing, dashboards, online evals |

Tracing runs through all of it. You will see a trace in every notebook.

## Under the hood

**Why a lifecycle at all.**

<img src="../../assets/slides/adlc-1-new-kind-of-software.jpg" width="720">

Agents are a new kind of software. Inputs are very wide: passengers send any photo and
any sentence. Outputs are non-deterministic and effectively infinite. You don't know what
your agent will do until it's in production.

<img src="../../assets/slides/adlc-2-iterate-systematically.jpg" width="720">

So the teams shipping the best agents iterate systematically. They don't tweak a prompt
and hope; they measure each change.

<img src="../../assets/slides/adlc-3-agent-engineering.jpg" width="720">

That makes agent engineering its own discipline: product thinking (user needs and
outcomes), engineering (systems, reliability, scale), and data science (evaluation
driven development). It is the iterative process of refining agents into reliable
experiences.

<img src="../../assets/slides/adlc-4-lifecycle.jpg" width="720">

The loop is Build, Test, Deploy, Monitor, with Govern around it. What you learn in
Monitor feeds the next Build.

**What connects the stages.**

<img src="../../assets/slides/adlc-6-traces.jpg" width="720">

Traces sit in the middle. Building produces them. Testing turns them into datasets and
scores them. Deployed agents emit them. Monitoring reads them, and the bad ones become
the next test cases. That is why tracing is the first LangSmith feature you turn on.

Docs: [LangSmith evaluation](https://docs.langchain.com/langsmith/evaluation).

## Recap

- Agent inputs and outputs are open-ended, so you can't know behavior before production.
- The lifecycle is Build, Test, Deploy, Monitor, with Govern around it.
- Traces connect every stage.
- Part 1 builds and tests a workflow, Part 2 an agent, Part 3 deploys and monitors it.
